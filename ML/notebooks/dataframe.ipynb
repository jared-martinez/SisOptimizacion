````python
#Importacion de librerias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
````

````python
#Crear dataframe
encoding = "latin1"
path = "../data/Obesidad_Infantil.csv"
df = pd.read_csv(path, encoding=encoding)
print(df)
````

## Dataframe

````python
#Explorar algunos atributos y métodos del df
print(df.shape)   # (filas, columnas) Devuelve una tupla con el número de rxc
print(df.info())  # Devuelve informacion del tipo de dato almacenado en cada columna
print(df.head())  # Por defecto las primeras 5 muestras (n)
print(df.tail())  # Por defecto las últimas 5 muestras (n)
````

````python
print(type(df))        #Me indica el tipo de instancia
print(df.describe())   #Devuelve las medidas descriptivas de cada variable
````

## Cargar el modelo ya entrenado

````python
#Cargar el modelo de regresion multiple guardado
ruta_modelo = "../models/modelo_regresion_multiple.joblib"
data = joblib.load(ruta_modelo)

modelo = data["model"]
metrics = data["metrics"]
feature_names = data["feature_names"]
X_test = data["X_test"]
y_test = data["y_test"]
y_pred = data["y_pred"]

print("Variables predictoras:", feature_names)
print("Metricas guardadas:", metrics)
print("Coeficientes:", modelo.coef_)
print("Intercepto:", modelo.intercept_)
````

````python
#Verificar las metricas recalculandolas sobre el set de prueba guardado
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)

print(f"R2:  {r2:.4f}")
print(f"MAE: {mae:.4f}")
print(f"MSE: {mse:.4f}")
````

## Gráfica: valores reales vs. predichos

````python
plt.figure(figsize=(7,7))
sns.scatterplot(x=y_test, y=y_pred, alpha=0.7)
lims = [min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())]
plt.plot(lims, lims, 'r--', label="Predicción perfecta")
plt.xlabel("Peso real (KG)")
plt.ylabel("Peso predicho (KG)")
plt.title("Modelo de regresión múltiple: Real vs Predicho")
plt.legend()
plt.tight_layout()
plt.savefig("../graphics/real_vs_predicho.png", dpi=150)
plt.show()
````
