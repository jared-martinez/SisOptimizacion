```python
#Importación de Librerias
from pyomo.environ import *
from pyomo.opt import SolverFactory
```

```python
#Crear el modelo
model=ConcreteModel(name="Gutchi-company")
```

```python
#Pasos para solucionar el modelo
#1. Definir las variables de decision
model.x1=Var(doc="Cantidad de bolsas de mano",domain=NonNegativeReals)
model.x2=Var(doc="Cantidad de bolsas para rasuradora",domain=NonNegativeReals)
model.x3=Var(doc="Cantidad de mochilas",domain=NonNegativeReals)
```

```python
#2. Definir la funcion objetivo
model.obj=Objective(
    expr=24*model.x1 + 22*model.x2 + 45*model.x3,
    sense=maximize,
    doc="Maximizar Ingresos totales"
)
```

```python
#3. Restricciones del modelo

#Restriccion de piel
model.piel=Constraint(
    doc="Disponibilidad de pies2 de piel",
    expr=2*model.x1 + model.x2 + 3*model.x3 <= 42
)

#Restriccion de costura
model.costura=Constraint(
    doc="Disponibilidad de horas de costura",
    expr=2*model.x1 + model.x2 + 2*model.x3 <= 40
)
#Restriccion de Terminado
model.terminado=Constraint(
    doc="Disponibilidad de horas de terminado",
    expr=model.x1 + 0.5*model.x2 + model.x3 <= 45
)
```

```python
#4. Resolver el modelo mediante el uso de un algoritmo (solver)
solver=SolverFactory("glpk")        #solver gratuito, una buena opcion es gurobi
results=solver.solve(model, tee= True)

print(results)
```
